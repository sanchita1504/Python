```python
# STARS PYRAMID

rows=int(input("Enter the no. of rows: "))
for i in range(rows): 
    for j in range(rows-i-1):
        print(" ",end="")
    
    for k in range(2*i+1):
        print("*",end="")

    print()
```

```python
#REVERSE TRIANGLE
n = 5
for i in range(n, 0, -1):
    stars = '*' * (2 * i - 1)
    print(stars.center(2 * n - 1))
```

```python
n = 5 
mid = n // 2

for i in range(n):
    for j in range(n):
        if i == mid or j == mid:
            print('*', end=' ')
        else:
            print(' ', end=' ')
    print()
```

```python
 #INCREASING NO. STYLE FULL PYRAMID

rows=int(input("Enter the no. of rows: "))
for i in range(1,rows+1): 
    for j in range(rows-i):
        print(" ",end="")
    
    for k in range(1,2*i):
        print(k,end="")

    print()
```

```python
#REPEATING NO. STYLE PYRAMID

rows= int(input("Enter the no. of rows: "))

for i in range (1,rows+1):
    for j in range(rows-i):
        print(" ",end="")
    for k in range(1,2*i):
        print(i,end="")

    print()
```

```python
# PALINDROME FULL PYRAMID

rows=int(input("Enter the no. of rows :"))

for i in range(1,rows+1):
    for j in range(rows-i):
        print(" ",end="")
    
    for j in range(i,0,-1):
        print(j,end="")
    
    for j in range(2,i+1):
        print(j, end="")
    
    print()
```

```python
#RECTANGULAR STAR PATTERN

rows = int(input("Enter the number of rows: "))
cols = int(input("Enter the number of columns: "))

for i in range(rows):
    for j in range(cols):
        print("*", end="")
    print()
```

```python
#RIGHT ANGLED TRIANGLE

rows = int(input("Enter the number of rows: "))

for i in range(1, rows + 1):
    for j in range(i):
        print("*", end="")
    print()
```

```python
#RIGHT ANGLED NUMBER TRIANGLE

rows = int(input("Enter the number of rows: "))

for i in range(1, rows + 1):
    for j in range(i):
        print(i, end="")
    print()
```

```python
#FLOYD'S TRIANGLE

rows = int(input("Enter the number of rows: "))

num = 1

for i in range(1, rows + 1):
    for j in range(i):
        print(num, end=" ")
        num += 1
    print()
```

```python
#INCREASING NUMBER IN EVERY ROW

rows = int(input("Enter the number of rows: "))

for i in range(1, rows + 1):
    for j in range(1, i + 1):
        print(j, end="")
    print()
```

```python
#INVERTED INCREASING NUMBER RIGHT PYRAMID

rows = int(input("Enter the number of rows: "))

for i in range(rows, 0, -1):
    for j in range(1, i + 1):
        print(j, end="")
    print()
```

```python
#DIAMOND STAR PATTERN

rows = int(input("Enter the number of rows: "))


for i in range(1, rows + 1):

    
    for j in range(rows - i):
        print(" ", end="")

    
    for k in range(2 * i - 1):
        print("*", end="")

    print()


for i in range(rows - 1, 0, -1):

    
    for j in range(rows - i):
        print(" ", end="")

    
    for k in range(2 * i - 1):
        print("*", end="")

    print()
```

```python
#HALF DIAMOND STAR PATTERN

rows = int(input("Enter the number of rows: "))


for i in range(1, rows + 1):
    for j in range(i):
        print("*", end="")
    print()


for i in range(rows - 1, 0, -1):
    for j in range(i):
        print("*", end="")
    print()
```

```python
#NUMBER CROWN PATTERN

rows = int(input("Enter the number of rows: "))

for i in range(1, rows + 1):

    
    for j in range(1, i + 1):
        print(j, end="")

    
    for j in range(2 * (rows - i)):
        print(" ", end="")

    
    for j in range(i, 0, -1):
        print(j, end="")

    print()
```

```python
#BUTTERFLY PATTERN

rows = int(input("Enter the number of rows: "))

for i in range(1, rows + 1):

    
    for j in range(i):
        print("*", end="")

    
    for j in range(2 * (rows - i)):
        print(" ", end="")

    
    for j in range(i):
        print("*", end="")

    print()


for i in range(rows - 1, 0, -1):

    
    for j in range(i):
        print("*", end="")

    
    for j in range(2 * (rows - i)):
        print(" ", end="")

    
    for j in range(i):
        print("*", end="")

    print()
```

```python
n = 7  
mid = n // 2

for i in range(n):
    row = ''
    for j in range(n):
        if i == 0 or i == mid or i == n - 1:          # top, middle, bottom bar
            row += '*'
        elif i < mid and j == 0:                       # left side, top half
            row += '*'
        elif i > mid and j == n - 1:                    
            row += '*'
        else:
            row += ' '
    print(row)
```

```python
# CHRISTMAS TREE
sections = 3
height = 5

for s in range(sections):

    for i in range(height):
        spaces = sections * height - i - height // 2 - s
        stars = 2 * (i + s + 1) - 1

        print(" " * spaces + "*" * stars)

for i in range(4):
    print(" " * (sections * height - 2) + "|||")
```

```python
# HOUSE
roof_height = 10
house_width = 21
wall_height = 6


for i in range(roof_height):
    spaces = roof_height - i - 1
    stars = 2 * i + 1

    print(" " * spaces + "*" * stars)


for i in range(wall_height):

    if i == wall_height - 1:
        print("|" + "_" * (house_width - 2) + "|")

    elif 1 <= i <= 3:
        left_space = (house_width - 9) // 2

        print(
            "|" +
            " " * left_space +
            "|" + " " * 5 + "|" +
            " " * left_space +
            "|"
        )

    else:
        print("|" + " " * (house_width - 2) + "|")
```
